```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.countplot(data=df,y="produto")
plt.title ("Produtos mais vendidos")
```

```python
df.isna ().sum()
```

```python
df["total"] = df["quantidade"] * df["preco_unitario"]
df.head()
```

```python
plt.title ("Faturamento por dia")
faturamento_por_dia = df.groupby ("dia_semana")["total"].sum().reset_index()
sns.barplot(data=faturamento_por_dia,x="total",y="dia_semana")
plt.show()
```
